# 1. IndentationError:

* Cause:
--- This error occurs when the indentation (spaces or tabs) in your code is not consistent.
Python relies on indentation to define the blocks of code (e.g., inside loops, functions, conditionals), and inconsistent indentation will lead to errors.
* Fix:
--- Ensure that you use the same number of spaces or tabs for each level of indentation.
Stick to either spaces or tabs (PEP 8 recommends using 4 spaces per indentation level).

```python
# Example of IndentationError
if True:
print("Hello")  # This will cause IndentationError
```

```python
# Correcting the indentation
if True:
    print("Hello")  # This is correct
```

# 2. NameError:


* Cause:
--- This error occurs when you try to use a variable or function that has not been defined yet.
It can happen when you mistype a variable or use a name that doesn't exist in the scope.
* Fix:
--- Ensure that all variables and functions are defined before they are used.
Double-check for typos in variable or function names.

```python
# Example of NameError
print(x)  # x is not defined
```

```python

# Fixing the error
x = 10
print(x)  # Output will be 10
```

# 3. ValueError:

* Cause:
This error occurs when a function receives an argument of the correct type but an inappropriate value.
For example, trying to convert a non-numeric string to an integer.
* Fix:
Check the input values before using them in functions that expect a specific value type.

```python
# Example of ValueError
num = int("hello")  # Cannot convert a non-numeric string to an integer
```

```python
# Fixing the error
try:
    num = int("123")
    print(num)  # Output will be 123
except ValueError:
    print("Invalid input, unable to convert.")
```

# 5. IndexError:

* Cause:
This error occurs when you try to access an index in a sequence (like a list or tuple) that is out of range.
For example, trying to access the 5th element in a list of only 3 elements.
* Fix:
Ensure that the index is within the valid range (0 to length of the list - 1).
You can also use conditionals to check if the index is valid.

```python
# Example of IndexError
my_list = [1, 2, 3]
print(my_list[5])  # Index out of range
```

```python

# Fixing the error
if len(my_list) > 5:
    print(my_list[5])
else:
    print("Index out of range")
```

# 6. KeyError:

* Cause:
This error occurs when you try to access a dictionary key that does not exist.
For example, trying to access a key that has not been added to the dictionary.
* Fix:
Use the in keyword to check if a key exists in the dictionary before accessing it.
Alternatively, use the get() method, which returns None (or a default value) if the key is not found.

```python
# Example of KeyError
student = {"name": "Alice", "age": 20}
print(student["gender"])  # 'gender' key does not exist
```

```python
# Fixing the error with `in`
if "gender" in student:
    print(student["gender"])
else:
    print("Key not found.")

# Using get() to handle KeyError
gender = student.get("gender", "Not Provided")
print(gender)  # Output will be 'Not Provided'
```
