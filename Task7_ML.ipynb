#### Handling Missing Values
* For numerical columns - you can fill missing values with the median or mean.
* For categorical columns - you can fill missing values with the mode.

```python
df['age'].fillna(df['age'].median(), inplace=True)
df['embark_town'].fillna(df['embark_town'].mode()[0], inplace=True)
df['deck'] = df['deck'].cat.add_categories('Unknown')  # Add 'Unknown' as a category
df['deck'].fillna('Unknown', inplace=True)
df['embarked'].fillna(df['embarked'].mode()[0], inplace=True)
```

## Label encoding

```python
label = LabelEncoder()
cate_col = df.select_dtypes(include = ["object"]).columns

for col in cate_col:
    df[col] = label.fit_transform(df[col])

print(df.head())
```

```python
label = LabelEncoder()
cate_col = df.select_dtypes(include = ["category"]).columns

for col in cate_col:
    df[col] = label.fit_transform(df[col])

print(df.head())
```
