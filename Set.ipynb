# Set

* Unordered: Sets do not maintain the order of elements.
* Unique Elements: Sets automatically remove duplicate elements.
* Mutable: You can add and remove elements, but sets cannot contain mutable elements like lists or dictionaries.
* No Indexing: Sets do not support indexing, slicing, or other sequence-like behavior.
* Mathematical Operations: Sets support operations like union, intersection, and difference.

```python
# Create sets
set_a = {10, 20, 30, 40}
set_b = {30, 40, 50, 60}

# 1. Adding an element
set_a.add(70)
print("After adding 70:", set_a)

# 2. Removing an element (using remove)
set_a.remove(20)
print("After removing 20:", set_a)

# 3. Discarding an element (does not raise error if element is not found)
set_a.discard(100)  # Element 100 does not exist in set_a
print("After discarding 100 (non-existing):", set_a)

# 4. Popping a random element
popped_element = set_a.pop()
print("Popped element:", popped_element)
print("Set after pop:", set_a)

# 5. Union of two sets
union_result = set_a.union(set_b)
print("Union of set_a and set_b:", union_result)

# 6. Intersection of two sets
intersection_result = set_a.intersection(set_b)
print("Intersection of set_a and set_b:", intersection_result)

# 7. Difference between two sets
difference_result = set_a.difference(set_b)
print("Difference of set_a and set_b:", difference_result)

# 8. Clear all elements from set_a
set_a.clear()
print("After clearing set_a:", set_a)
```
