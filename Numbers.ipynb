# Numbers in Python

* Integers
      --- Whole numbers positive negative ex: 43 -32
* Float
      --- Numbers with decimal points ex: 4.2  88.93
* Complex
       --- Numbers with real and imaginary values ( a+bj)
  i.e a is a real and b is a imaginary value
  ex: 7-4j   -8+2j

```python
a=54
print(type(a))
```

```python
b=7.9
print(type(b))
```

```python
c=3+4j
print(type(c))
```

### Arithmetic operations:


### 1. Addition '+' 

```python
a=2           # To calculate the sum
b=3
print(a+b)
```

```python
A=10     # int
B=2.5    # float
C=6+5j    # complex
D=-8+5j
print(A+B)
print(C+B)
print(A+C)
print(C+D)
```

### 2. Substraction '-'

```python
x=4                   # # To calculate the difference
y=83
print(x-y)
```

```python
A=10     # int
B=2.5    # float
C=6+5j    # complex

print(A-B)
print(C-B)
print(A-C)
print(C-D)
```

### 3. Multiplication  '*'

```python
a=2                        #  # To finding the product
b=3
print(a*b)
```

```python
A=10     # int
B=2.5    # float
C=6+5j    # complex

print(A*B)
print(C*B)
print(A*C)
print(C*D)
```

### 4. Division '/'

```python
x=4                            # To finding the quotient
y=83
print(y/x)
```

```python
A=10     # int
B=2.5    # float
C=6+5j    # complex

print(A/B)
print(C/B)
print(A/C)
print(C/D)
```

### 5. Floor division '//'

```python
x=4
y=83
print(y//x)
```

### 6. Exponentiation '**'

```python
a= 7**2
print(a)
```

### 7. Module '%'

```python
c= 47 % 8
print(c)                            # To finding the Reminder
```
