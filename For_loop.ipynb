# For Loop

 A for loop in python is used to iterate over a sequence (like a list, tuple, string or range) and execute a block of code multiple times. 
It's a powerful control flow tool, especially useful when you konw the number of iterations in advance or want to process items in a sequence. 

### 1. Print odd values between 20 and 80 without using if, using only a for loop

```python
for i in range(21, 80, 2):  # increment by 2
    print(i)
```

### 2. Create a list of numbers from 1 to 20 using a for loop

```python
numbers = []
for i in range(1, 21):  # Loop from 1 to 20
    numbers.append(i)
print(numbers)
```

### 3. Create a list of numbers from 20 to 1 using a for loop (without reverse)

```python
numbers = []
for i in range(20,0,-1):  # Loop from 1 to 20
    numbers.append(i)
print(numbers)
```

### 4. Take the cube of odd values between 20 to 40

```python

for i in range(21, 40, 2):  # Start at 21, go up to 40, increment by 2 (only odd numbers)
    print(f"Cube of {i} is {i**3}")
```

### 5. Combine names and ages from two lists and print them

```python
names = ["Harshad", "Soha", "priyag", "Roshni", "Diya"]
ages = [25, 22, 30, 24, 32]

for i in range(len(names)): 
    print(f"my name is {names[i]} , my age is {ages[i]}")
```

### 1. Swap First and Last Elements of a List

```python
Numbers_list = [10, 20, 30, 40, 50]

if len(Numbers_list) > 1:
    variable = Numbers_list[0]  
    Numbers_list[0] = Numbers_list[-1]  # Put 50 at the first position
    Numbers_list[-1]=variable
    
print("List :", Numbers_list)
```

### 2. Count the Length of a String

```python
str = "I am from kolhapur"

length = 0
for char in str:
    length += 1

print("Length of the string:", length)
```

### 3. Sum of Non-Negative Integers in a List

```python
list = [-2, 4, -3, -5, -20, 10]

Sum = 0
for i in list:
    if i >= 0:
        Sum += i

print("Sum of non-negative integers:", Sum)
```

### 4 write code of factorial

```python
Number = 6

factorial = 1
for i in range(1, Number+1):
    factorial *= i

print(f" Factorial of {Number} is:", factorial)
```
