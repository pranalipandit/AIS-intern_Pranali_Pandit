# 1. Arithmetic Operators

```python
a = 43
b = 51
print("Addition:", a + b)       
print("Subtraction:", a - b)    
print("Multiplication:", a * b) 
print("Division:", a / b)       
```

# 2. Comparison Operators

```python

These operators compare values and return a Boolean result (True or False).
 ==, !=, <, >, <=, >=

Operator	Description	Example
==      	Equal to	
!=      	Not equal to	
<       	Less than	
>	        Greater than	
<=	        Less than or equal to	
>=	        Greater than or equal to	
```

```python
x = 37
y = 10
print(x == y) 
print(x != y)  
print(x < y)   
print(x > y)   
print(x <= y)   
print(x >= y)   
```

# 3. Logical Operators

```python
salary = 35000
print(salary > 10000 and salary < 40000)  
print(salary > 20000 or salary < 15000)   
print(not (salary  > 25000))         
```

# 4. Assignment Operators

```python
a = 15
a += 5  # a = a + 5
print(a) 

b = 32
b -= 3
print(b)

c = 5
c *= 4
print(c)

x = 4
x **= 2  # x = x ** 2
print(x)  

y = 2
y /= 6
print(y)

y = 2
y //= 6
print(y)

b = 32
b %= 3
print(b)
```
