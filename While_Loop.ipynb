# While Loop

   * A while loop is a control flow statement in programming languages that repeats a section of code until a condition is met, 
It's similat to a repeating if statement.

### 1. Odd-Even Numbers Using While Loop

```python
n = 1

while n <= 12 :
    if n % 2 == 0:
        print(f"{n} is even")
    else:
        print(f"{n} is odd")
    n += 1
```

### 2  using while loop print pattern

```python
rows = 5
i = 1

while i <= rows:
    print(" * " * i)
    i += 1
```

### 3.create list 1-20 numbers list using while loop=> [1,2,3... 20]

```python
num = 1
numbers = []

while num <= 20:
    numbers.append(num)
    num += 1

print(numbers)
   
```

### 4. creat list 20-1 (revers order) using while loop=> [20,19...1]

```python
num =20
numbers = [] 

while num >= 1:
    numbers.append(num)
    num -= 1
print(numbers)
```

### 5. try with one any eg. break, continue , pass control statement

```python
#1 Break

num =1
while num <=15:
    if num == 6:
        print("The loop is break down")
        break
    print(num)
    num += 1
    
```

```python
#2 continue

num =1
while num <= 20:
    if num == 3:                    # skip 3 and continue rest of the loop
        num += 1
        continue
    print(num)
    num +=1
 
```

```python
# 3.pass:

num =1
while num <= 10:
    if num == 5:
        pass
    print(num)
    num += 1
```
