# Tuple

* Immutable: Elements cannot be changed after creation.
* Ordered: Elements maintain their order.
* Indexing: Access elements using positive or negative indices.
* Slicing: Extract portions of the tuple using slicing (tuple[start:end]).
* Heterogeneous: Can store different data types in the same tuple.
* Concatenation: Combine tuples using +.
* Repetition: Repeat tuples using *.
* Count: Use count() to count occurrences of an element.
* Index: Use index() to find the first occurrence index of an element.
* Length: Use len() to find the number of elements.
* Nested Tuples: Tuples can contain other tuples.
* Memory Efficient: More memory-efficient than lists due to immutability.
* Unpacking: Elements can be unpacked into variables.
* Used for Data Integrity: Ideal for cases where data should remain constant.








```python
# Create a tuple
fruits = ("apple", "banana", "cherry", "apple")

# 1. Accessing Elements using Indexing
print("First element:", fruits[0])  
print("Last element:", fruits[-1]) 

# 2. Slicing the Tuple
print("Sliced tuple (1 to 3):", fruits[1:3])  

# 3. Concatenation of Tuples
more_fruits = ("orange", "grape")
combined_fruits = fruits + more_fruits
print("Concatenated tuple:", combined_fruits)  

# 4. Repetition of a Tuple
repeated_fruits = fruits * 2
print("Repeated tuple:", repeated_fruits) 

# 5. Counting Occurrences of an Element
apple_count = fruits.count("apple")
print("'apple' occurs", apple_count, "times in the tuple.") 

# 6. Finding the Index of an Element
banana_index = fruits.index("banana")
print("'banana' is at index:", banana_index) 

# 7. Length of the Tuple
tuple_length = len(fruits)
print("Length of the tuple:", tuple_length) 

# 8. Nested Tuple
nested_tuple = (1, (2, 3), "apple")
print("Nested tuple:", nested_tuple)  # (1, (2, 3), 'apple')

# 9. Unpacking the Tuple
a, b, c, d = fruits
print("Unpacked values:", a, b, c, d)  # apple banana cherry apple
```
