# Dictionaries in python

* Unordered: Dictionaries are unordered collections of key-value pairs.
* Mutable: Dictionaries can be modified by adding, removing, or changing key-value pairs.
* Unique Keys: Each key in a dictionary must be unique, but values can be duplicated.
* Access by Keys: Values are accessed using keys, not indices.
* No Indexing: Unlike lists and tuples, dictionaries do not support indexing, slicing, or other sequence-like operations.
* Flexible: The keys and values can be of any data type (strings, integers, lists, etc.).

```python
# Create a dictionary
student = { "name": "John", "age": 25, "course": "Math"}
print(student)

# 1. Accessing a value using a key
print("Name:", student["name"])  

# 2. Accessing a value using the get() method
print("Age:", student.get("age")) 
print("Gender:", student.get("gender", "Not Found"))  

# 3. Adding a new key-value pair
student["address"] = "123 Street Name"
print("Updated dictionary:", student)

# 4. Removing a key-value pair using pop()
removed_value = student.pop("course")
print("Removed 'course':", removed_value)
print("After pop:", student)

# 5. Removing an arbitrary key-value pair using popitem()
popped_item = student.popitem()
print("Popped item:", popped_item)
print("After popitem:", student)

# 6. Accessing keys, values, and items
print("Keys:", student.keys())  # dict_keys(['name', 'age', 'address'])
print("Values:", student.values())  # dict_values(['John', 25, '123 Street Name'])
print("Items:", student.items())  # dict_items([('name', 'John'), ('age', 25), ('address', '123 Street Name')])

# 7. Updating a dictionary
student.update({"age": 26, "city": "New York"})
print("After update:", student)

```
