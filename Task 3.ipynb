### 4)Write a Python function total number of  Combinations. 

Ex. a,b,c
Ans: 3 : ab,ac,bc


```python
def total_combination(elements):
    elements = list(elements)
    combinatin_list = []
    for i in range (len(elements)):
        for j in range( i+ 1, len(elements)):
            combinatin_list.append(elements[i] + elements[j])
    return combinatin_list
```

```python
input_elements = "abc"
result = total_combination(input_elements)
print("Total_combinations:", result)
```

### 5.Write a Python function total number of  permutation 
Ex. a,b,c
And: 6 : ab,ac,ba,bc,ca,cb 


```python
def total_permutation(elements):
    elements = list(elements)
    permutation_list = []
    for i in range(len(elements)):
        for j in range(len(elements)):
            if i != j:
                permutation_list.append(elements[i] + elements[j])
    return permutation_list
```

```python
elements = "abc"
result = total_permutation(elements)
print("Total_permutations:", result)
```

### 6 Define a function which counts vowels and consonant in a word.
  input :  statistics 
Expected output : Count of vowel is = 3 Count of consonant is = 7


```python
def count_vowels_constants(input):
    vowels ="aeiou"
    vowel_count = 0
    constant_count = 0
    
    for char in input.lower():
        if char in vowels:
             vowel_count += 1
        elif char.isalpha():
            constant_count += 1
            
    print(f"count of vowels is = {vowel_count}")
    print(f"count of constant is = {constant_count}")
```

```python
count_vowels_constants("statistics")
```

### 8. count lower case and upper case  letter.
Ex : STatiStiCS
Ans: upper case :5
Lower case: 5

```python
def count_case(word):
    lower_case = 0
    upper_case = 0
    for char in word:
        if char.islower():
            lower_case += 1
        elif char.isupper():
            upper_case +=1

    print(f"count of lowercase letter is = {lower_case}")
    print(f"count of uppercase latter is = {upper_case}")
```

```python
count_case("STatIStiCS")
```
