## Seaborn Plot types
Seaborn provides a wide range of plot types that can be used for data visualization and exploratory data analysis. Broadly speaking, any visualization can fall into one of the three categories. 

* Univariate : x only (contains only one axis of information)
* Bivariate : x and y (contains two axis of information)
* Trivariate : x, y, z (contains three axis of information)


### 1.Scatter Plot. 
A scatter plot is used to visualize the relationship between two variables. Seaborn's scatterplot() function provides a simple way to create scatter plots.

```python
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")
sns.scatterplot(x="total_bill", y= "tip", hue="sex", size = "size", sizes=(50,100),data=tips)

plt.xlabel("Total bill")
plt.ylabel("Tip")
plt.title("Relationship between Total bill and Tip")
plt.show()
```

#### Interpretation:
The scatter plot shows a positive correlation between total bill and tip amount, with larger bills generally receiving larger tips.  
Point size indicates party size, and color distinguishes sex, allowing for visual comparison of tipping behavior across these categories.

### 2.Line plot:
A line plot is used to visualize the trend of a variable over time.
Seaborn's lineplot() function provides a simple way to create line plots.

```python
flights = sns.load_dataset("flights")
flights
```

```python
sns.lineplot(x="year", y="passengers",data=flights)
```

```python
flights = sns.load_dataset("flights")
sns.lineplot(x="year", y="passengers", hue="month",data=flights)
plt.xlabel("Year")
plt.ylabel("Passengers")
plt.title("Changes in passengers numbers over time")
plt.show()
```

#### Interpretation:
This line plot shows a clear upward trend in the number of airline passengers over time (from 1949 to 1960).  The different colored lines represent the months, revealing a consistent seasonal pattern with peaks in passenger numbers during the summer months (particularly July and August) each year.

### 3.Bar plot:
* Bar plots are used to visualize the relationship between a categorical variable and a continuous variable. 
* In a bar plot, each bar represents the mean or median (or any aggregation) of the continuous variable for each category. 
* In Seaborn, bar plots can be created using the barplot() function. 

```python
iris = sns.load_dataset("iris")
iris
sns.barplot(x='species',y='petal_length',data=iris)
```

```python
iris
```

```python
iris = sns.load_dataset("iris")
sns.barplot(x='species',y='petal_length',data=iris,palette='viridis')

plt.xlabel("Species")
plt.ylabel("Petal length (cm)")
plt.title("Petal Length by Species")

plt.show()
```

#### Interpretation
This bar plot displays the average petal length for each species of iris flower (setosa, versicolor, and virginica).  
It clearly shows that setosa has the shortest petal length, followed by versicolor, and then virginica with the longest petal length.

```python

titanic = sns.load_dataset("titanic")
sns.barplot(x="class", y="fare", hue="sex", ci=None, palette="muted", data=titanic)

plt.xlabel("Class")
plt.ylabel("Fare")
plt.title("Average Fare by Class and Gender on the Titanic")
plt.show()
```

#### Interpretation
The bar plot visualizes average Titanic fares by class and gender, revealing higher fares for higher classes and slightly higher fares for men 
within each class.

### 4.Histogram:
* A histogrram is used to visualize the distribution of a variable.
* Seaborn's histplot() fuction provide a simple way to create histograms.

```python
iris = sns.load_dataset('iris')
sns.histplot(x='petal_length',data=iris)
```

```python
iris = sns.load_dataset('iris')
sns.histplot(x='petal_length',bins=20,data=iris,color='r')
plt.xlabel('Petal length')
plt.ylabel('Frequency')
plt.title('Distribution of Petal Lengths in Iris Flowers')
plt.show()
```

#### Interpretation
This histogram displays the distribution of petal lengths for Iris flowers, showing the frequency of flowers within 20 petal length bins. 
The red color highlights the distribution's shape, revealing potential clusters or patterns in petal size.

### 5.Density plot:
Density plot is also known as kernal density plot are type of visuallization that display the variation of continueous variabe

```python

tips = sns.load_dataset("tips")

sns.kdeplot(x='total_bill',data=tips)
```

```python
tips = sns.load_dataset("tips")

sns.kdeplot(x='total_bill',data=tips,hue='time',fill=True, alpha=0.6, linewidth=1.5)
```

#### Interpretation:
This KDE plot visualizes the distribution of total bills, separated by time of day (Lunch and Dinner). The filled areas and transparency allow for easy comparison of the distributions, showing how the density and spread of total bills differ between lunch and dinner.

### 6. box plots:
A box plot is used to visualize the distribution of a variable. Seaborn's boxplot() function provides a simple way to create box plots.




```python
tips = sns.load_dataset("tips")

sns.boxplot(x="day", y="total_bill", data=tips)
```

```python
tips = sns.load_dataset("tips")

sns.boxplot(x="day", y="total_bill", hue="time", data=tips, palette="Set3", linewidth=1.5, fliersize=4)

plt.title("Box Plot of Total Bill by Day and Meal Time")
plt.xlabel("Day of the Week")
plt.ylabel("Total Bill ($)")

plt.show()
```

#### Interpretation:
The box plot compares total bill distributions across days and meal times, showing variations in median, quartiles, and outliers to reveal how bill amounts differ across these categories.

### 7.violin plot
A violin plot is similar to a box plot, but provides a more detailed view of the distribution of the data. Seaborn's violinplot() function provides a simple way to create violin plots.

```python
iris = sns.load_dataset("iris")
sns.violinplot(x="species", y="petal_length", data=iris)
plt.show()
```

#### Interpretation:
The violin plot displays the distribution of petal lengths for each iris species, combining box plot summaries with kernel density estimations
for a detailed view of petal length variation.

### 8.pair plots:
 A pairplot is used to visualize the relationship between multiple variables. Seaborn's pairplot() function provides a simple way to create pairplots.

```python
iris = sns.load_dataset("iris")

sns.pairplot(data=iris, hue="species", diag_kind="kde", palette="husl")
plt.title("Iris Dataset Pair Plot")

plt.show()
```

#### Interpretation:
The pair plot visualizes all pairwise relationships and individual distributions of features in the Iris dataset, colored by species, 
to reveal correlations and species separation.

### 9.Joint Plot
The jointplot is a powerful visualization tool in Seaborn that allows you to examine the relationship between two variables while simultaneously 
observing the distribution of each individual variable.

```python
tips = sns.load_dataset('tips')

# Create a jointplot
sns.jointplot(x='total_bill', y='tip', data=tips, kind='hex')  
plt.show()
```

#### Interpretation:
The pair plot visualizes all pairwise relationships and individual distributions of features in the Iris dataset, colored by species, to reveal correlations and species separation.
