# 3. Array Operation

#### 1. Arithmatic Operations on numpy Arrays
    Numpy supports element-wise arithmatic between arrays of the same shape. 

```python
import numpy as np

a = np.array([4,6,8])
b = np.array([3,6,2])

#element-wise arithmatic operations
addition_1 = a + b

substraction_2 = a - b

multiplication_3 = a * b

division_4 = a / b

power_a = a ** 2

power_b = b ** 2
print("sum:", addition_1)
print("difference:", substraction_2)
print("product:", multiplication_3)
print("division:", division_4)
print("power of a:", power_a)
print("power of b:", power_b)
```

#### 2.Broadcasting in NumPy

    Broadcasting is a powerful mechanism in NumPy that allows operations between arrays of different shapes. It enables efficient computations without the need to explicitly create intermediate arrays, leading to significant performance gains.

* Key Rules of Broadcasting:

Rule 1: Shape Compatibility

Arrays must have compatible shapes for broadcasting to occur.
Two dimensions are compatible when:
* They are equal.
* One of them is 1.
  
Rule 2: Implicit Expansion

Dimensions of size 1 are stretched or "broadcast" to match the corresponding dimension of the other array.

```python
import numpy as np

a = np.array([1, 2, 3])  # Shape: (3,)
b = np.array([4])       # Shape: (1,)

result = a + b  
print("Array brodcasting is:", result)
```

```python
a = np.array([1, 2, 3])
b = 2

result = a * b  
print("Scalar Broadcasting:", result)
```

```python
A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([1, 2, 3])  # B is broadcasted across rows

result = A + B
print("Array Broadcasting:\n", result)
```

#### 3.Element-wise Operations and Functions
    Element-wise operations and functions in NumPy operate on each element of an array independently. This is often more efficient than using loops in Python, especially for large arrays.

```python
import numpy as np

a = np.array([1, 2, 3])

result_sqrt = np.sqrt(a)
result_sin = np.sin(a)    
result_cos = np.cos(a)
result_exp = np.exp(a)
result_log = np.exp(a)
result_abs = np.abs(a)

print("square root of a :", result_sqrt)
print("sine of a:", result_sin)
print("cosine of a:", result_cos)
print("exponential of a:", result_exp)
print("natural logarithm of a:", result_log)
print("absolute value of a:", result_abs )
```

# 4. Array Shapes and Reshaping

#### 1.Checking array dimensions with .shape and .ndim.

##### .shape :
        Returns a tuple representing the dimensions of the array.

```python
import numpy as np

arr = np.array([[3, 3, 5], [5, 3, 2]])
print(arr.shape)
```

```python
arr = np.array([[[3, 3, 5], [5, 3, 2], [7, 4, 2]]])
print(arr.shape)
```

#### .ndim: 
    Returns the number of dimensions (or axes) of the array.

```python
arr = np.array([[3, 3, 5], [5, 3, 2]])
print(arr.ndim)

arr = np.array([[[3, 3, 5], [5, 3, 2], [7, 4, 2]]])
print(arr.ndim)
```

#### 2.Changing Array Shape with  .reshape(), .flatten(), .transpose().

#### .reshape()
    Reshapes the array to a new shape while keeping the same data.

```python
arr = np.array([3, 2, 7, 6, 2, 4])
new_arr = arr.reshape(3,2)
print(new_arr)
```

```python
arr = np.array([3, 2, 7, 6, 2, 4])
new_arr = arr.reshape(2,3)
print(new_arr)
```

#### .flatten():
    Converts the array into a 1D array by flattening all dimensions.

```python
arr = np.array([[2,4], [1,2]])
flat_arr = arr.flatten()
print(flat_arr)
```

```python
arr = np.array([[[2,4], [1,2],[6,3]]])
flat_arr = arr.flatten()
print(flat_arr)
```

#### .transpose():
    Switches the axes of the array. For a 2D array, it swaps rows and columns.

```python
arr = np.array([[1, 2], [3, 4]]) 
transposed_arr = arr.transpose() 
print(transposed_arr)
```

```python
arr = np.array([[[1, 2, 2], [4, 5, 1], [6,8,2]]])
transposed_arr = arr.transpose()
print(transposed_arr)
```

#### 3. Concatenating and splitting arrays using np.concatenate(), np.vstack(), np.hstack(), np.split().


#### np.concatenate() : 
Concatenates arrays along a specified axis.

```python
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])
concatenated_arr = np.concatenate((arr1, arr2), axis=0)  # Concatenate vertically
print(concatenated_arr)
```

```python
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])
concatenated_arr = np.concatenate((arr1, arr2), axis=1)  # Concatenate horizontally
print(concatenated_arr)
```

#### np.vstack():
Concatenates arrays vertically (row-wise).

```python
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])
vstack_arr = np.vstack((arr1, arr2))
print(vstack_arr)
```

#### np.hstack(): 
Concatenates arrays horizontally (column-wise).

```python
hstack_arr = np.hstack((arr1, arr2))
print(hstack_arr)
```

#### np.split():
Splits an array into multiple sub-arrays along a specified axis

```python
arr = np.array([1, 2, 3, 4, 5, 6])
split_arr = np.split(arr, 3)  # Split into 3 equal parts
print(split_arr)
```

```python
arr = np.array([1, 2, 3, 4, 5, 6])
split_arr = np.split(arr, 2)  # Split into  2 equal parts
print(split_arr)
```
