## Random Sampling, Shuffling, and Permutations

### np.random.choice(): 
Draws a given number of samples from a given array or sequence with or without replacement.

```python
data = np.array([2,4,2,1,5,6])
sample = np.random.choice(data, 7, replace = True) # replace=True allows sampling with duplicates
sample
```

```python

data = np.array([1, 2, 3, 4, 5])
samples = np.random.choice(data, 5, replace=False)  # replace=False don't allows sampling with duplicates
print(samples)
```

### np.random.shuffle():
Shuffles the elements of an array in-place.

```python
arr = np.array([1, 2, 3, 4, 5])
sample = np.random.shuffle(arr)
print(arr)
```

```python
data = np.array([12, 23, 13, 24, 51])
sample = np.random.shuffle(data)
print(data)
```

### np.random.permutation():
Returns a permuted copy of an array or a sequence of integers.

```python
data = np.array([1,  4, 5, 3, 8])
permutation_sample = np.random.permutation(data)
print(data)
```

```python
data = np.array([1, 2, 3, 4, 5])
permuted_data = np.random.permutation(data)
print(permuted_data)
```

# 8. Broadcasting

Broadcasting is a powerful mechanism in NumPy that allows operations between arrays of different shapes, as long as certain compatibility rules are met. It avoids the need to explicitly create intermediate arrays, making your code more concise and efficient.

Rules of broadcasting (e.g., matching dimensions, aligning axes).

Arrays must have compatible shapes along each dimension.
Dimensions are compatible when they are equal or one of them is 1.

Arrays are aligned based on their trailing (rightmost) dimensions.
Dimensions of size 1 are stretched or broadcast to match the corresponding dimension of the other array.

```python
arr1 = np.array([1,2,2])
arr2 = 4
print(arr1 * arr2)
```

```python
arr = np.array([4,3,2])
arr1 = np.array([[2,1,3], [4,2,3]])
print(arr + arr1)
```
