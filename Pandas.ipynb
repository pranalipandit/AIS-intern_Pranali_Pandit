### Aggregation: .sum(), .mean(), .count()
    Once you have your groups, you can perform operations (like summing, averaging, counting) on each group separately.

##### .sum()
sum() function adds up all the numerical values within each group.

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Values': [10, 20, 30, 40, 50, 60]
}

dfc = pd.DataFrame(data)

# Group by 'Category'
grouped = dfc.groupby('Category')

# Sum values within each group
sum_values = grouped.sum()
print(sum_values)
```

##### .mean()
The mean() function calculates the average of the numerical values within each group.

```python
mean_values = grouped.mean()           # Mean values within each group
print(mean_values)
```

##### .count()
The count() function counts the number of non-null entries within each group.

```python
count_values = grouped.count()         # Count of values within each group
print(count_values)
```

### Merging and Joining:
##### pd.merge()
pd.merge() is used for merging dataframes on specific columns or indexes. It resembles SQL joins (inner join, outer join, left join, right join) and gives you more control over the merging process.

```python
# Create two sample dataframes
df3 = pd.DataFrame({'key': ['A', 'B', 'C', 'D'],
                    'value1': [1, 2, 3, 4]})
df4 = pd.DataFrame({'key': ['B', 'D', 'E', 'F'],
                    'value2': [5, 6, 7, 8]})

# Perform an inner merge on the 'key' column
result = pd.merge(df3, df4, on='key', how='inner')          # inner - common keys

print(result)
```

```python
result = pd.merge(df3, df4, on='key', how='outer')           #outer- there is no value of key this is nan
print(result)
```

```python
result = pd.merge(df3, df4, on='key', how='left')           
print(result)
```

```python
result = pd.merge(df3, df4, on='key', how='right')
print(result)
```

##### pd.concat()
pd.concat() is used to concatenate dataframes along a particular axis (rows or columns). It's useful for appending dataframes or combining them along columns.

```python
df5 = pd.DataFrame({'A': ['A0', 'A1', 'A2'],
                    'B': ['B0', 'B1', 'B2']})
df6 = pd.DataFrame({'A': ['A3', 'A4', 'A5'],
                    'B': ['B3', 'B4', 'B5']})

# Concatenate dataframes along rows
result = pd.concat([df5, df6], axis=0)

print(result)
```

## Time series:
Time series analysis is a fascinating field that deals with data points collected or recorded at specific time intervals.

##### pd.to_datetime():
This function is used to convert a column to datetime.

```python
df = pd.DataFrame({
    'date': ['2025-01-01', '2025-01-02', '2025-01-03'],
    'value': [100, 200, 300]
})

# Convert 'date' to datetime
df['date'] = pd.to_datetime(df['date'])

# Set 'date' as the index
df.set_index('date', inplace=True)

print(df)
```

##### .resample()
This method is used to resample time series data. You can downsample (e.g., from daily to monthly) or upsample (e.g., from monthly to daily) using this method.

```python
# Create a sample dataframe
date_rng = pd.date_range(start='2025-01-01', end='2025-01-10', freq='D')
df = pd.DataFrame(date_rng, columns=['date'])
df['data'] = range(10)

# Set 'date' column as index
df.set_index('date', inplace=True)

# Resample to sum data by 3-day intervals
df_resampled = df.resample('3D').sum()

print(df_resampled)
```

### Condition Selection: And ,or
    you can use logical operators to select data based on multiple conditions. The & (and) and | (or) operators come in handy for condition selection.

##### & (And) Operator:
    The & operator is used to combine multiple conditions where all conditions must be true.

```python
# Sample data
df = pd.DataFrame({
    'A': [1, 2, 3, 4, 5],
    'B': [10, 20, 30, 40, 50]
})

# Select rows where A > 2 and B < 50
result = df[(df['A'] > 2) & (df['B'] < 50)]

print(result)
```

##### | (Or) Operator
    The | operator is used to combine multiple conditions where at least one condition must be true.

```python
# Select rows where A > 4 or B < 20
result = df[(df['A'] > 4) | (df['B'] < 20)]
print(result)
```

##### Combining Multiple Conditions
You can also combine multiple conditions using both & and | operators for more complex selections.

```python
# Select rows where A > 1 and (B < 30 or B > 40)
result = df[(df['A'] > 1) & ((df['B'] < 30) | (df['B'] > 40))]

print(result)
```
