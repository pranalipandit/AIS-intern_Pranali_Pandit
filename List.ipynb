# List

* Mutable: Lists can be modified after creation (items can be added, removed, or changed).
* Ordered: Lists maintain the order of elements.
* Indexed: Elements are accessed using indices (both positive and negative).
* Slicing: You can extract a portion of a list using slicing (list[start:end]).
* Heterogeneous: Lists can store different data types (integers, strings, etc.).
* Concatenation: Lists can be combined using the + operator.
* Repetition: Lists can be repeated using the * operator.
* Adding Elements: Use append(), insert(), or extend() to add items.
* Removing Elements: Use remove(), pop(), or clear() to remove items.
* Searching: Use count() to count occurrences of an element, and index() to find the index of an element.
* Sorting: Lists can be sorted using the sort() method.
* Reversing: You can reverse a list using the reverse() method.
* Length: Use len() to find the number of elements in a list.
* Nested Lists: Lists can contain other lists (nested lists).
* Unpacking: Elements of a list can be unpacked into variables.d into variables.


```python
# Step 1: Create a list of fruits
fruits = ["apple", "banana", "cherry"]
print("Fruits:", fruits)

# Step 2: Add more fruits
fruits.append("orange")  # Add at the end
fruits.insert(1, "mango")  # Add at index 1
print("\nAfter adding fruits:", fruits)

# Step 3: Access specific fruits
print("\nFirst fruit:", fruits[0])
print("Last fruit:", fruits[-1])

# Step 4: Change a fruit
fruits[2] = "grape"
print("\nAfter changing a fruit:", fruits)

# Step 5: Remove some fruits
fruits.remove("mango")  # Remove by name
print("\nAfter removing 'mango':", fruits)

popped_fruit = fruits.pop()  # Remove the last fruit
print("\nRemoved fruit:", popped_fruit)
print("Remaining fruits:", fruits)

# Step 6: Sort and reverse the fruits
fruits.sort()
print("\nSorted fruits:", fruits)

fruits.reverse()
print("Reversed fruits:", fruits)

# Step 7: Clear the fruit list
fruits.clear()
print("\nAll fruits cleared:", fruits)
```
