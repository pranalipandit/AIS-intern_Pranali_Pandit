### 1. if Statement
      The if statement is used to test a condition. If the condition evaluates to True, the block of code under the if statement is executed. If the condition is False, the block of code is skipped.


* Executes only if the condition is True.
* Can be used alone (no need for elif or else).


```python
# Example
```

```python
#1.check if a number is negative
num = 27
if num < 0:
    print(f"{num} is negative")

    
#2 check if a starting contains a specific word 
sentence = "This language is amazing"
if "amazing" in sentence:
    print("The sentence mentions amazing")

#3. check the file path ends with an .csv
file_name = "document.csv"
if file_name.endswith(".csv"):
    print("This is comma-seperated values file")


#4. check if a person is adult
age = 14
if age > 18:
    print("you are an adult")

#5.Check if Number is perfect square:
import math 
num = 36
if math.sqrt(num) == int(math.sqrt(num)):
    print("This is a perfect square")


```

### 2.  else statement

        The else statement in Python is used to define a block of code that will run if none of the previous if or elif conditions are True. It acts as a "default" or "fallback" condition and is always placed at the end of an if or if-elif chain.

* No Condition:
    Unlike if and elif, the else statement does not take a condition. It simply runs if all prior conditions are False.

* Optional:
    Using an else statement is not mandatory. You can have if and elif without an else.

* Placement:
    else must always come after all if and elif blocks.
    If else is placed before or outside the if chain, it will result in a syntax error.

* Execution:
    The code inside the else block will execute only when no prior condition in the if or elif chain evaluates to True.


```python
#1 To check a number is even or odd
num = 5
if num % 2 == 0:
    print(f"{num} is even")
else: 
    print(f"{num} is odd")

#2 Who is eligible to vote
age = 16
if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")


#3 To check a string is empty or not
str = "Hello"
if str:
    print("The string is not empty")
else:
    print("The string is empty")

#4 To check Temperature
temp = 40
if temp > 30:
    print("It's hot outside")
else: 
    print("It's not too hot")

#5 To check pass or fail
marks = 35
if marks >= 35:
    print("pass")
else:
    print("fail")
```

# 3.elif statement


    The elif (short for "else if") statement is used to check multiple conditions in a sequence. When the if condition is False, the program checks the elif conditions in order. As soon as one condition evaluates to True, the corresponding block of code is executed, and the rest of the elif conditions are ignored.


* Used to check additional conditions after an if.
* You can have multiple elif statements.
* Stops checking once a condition is satisfied.

```python
# 1 To Assign a grade based on marks
marks = 35
if marks > 75:
    print("Grade: O")
elif 60 < marks <= 75:
    print("Grade: A")
elif 50 < marks <= 60:
    print("Grade: B")
elif 35 <= marks <= 50:
    print("Grade: C")
else:
    print("Fail")

# 2 To Check number category
num = 3
if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")

# 3 To Identify a weekday
day = 6
if day == 1:
    print("Monday")
elif day == 2:
    print("Tuesday")
elif day == 3:
    print("Wednesday")
elif day == 4:
    print("Thursday")
elif day == 5:
    print("Friday")
else:
    print("Weekend")

# 4 To Check a person's age group
age = 15
if age <= 12:
    print("Child")
elif 13 <= age <= 19:
    print("Teenager")
elif 20 <= age <= 59:
    print("Adult")
else:
    print("Senior")

# 5 To Check whether a number is divisible by 2, 3, or both
n = 6
if n % 2 == 0 and n % 3 == 0:
    print("Divisible by both 2 and 3")
elif n % 2 == 0:
    print("Divisible by 2")
elif n % 3 == 0:
    print("Divisible by 3")
else:
    print("Not divisible by 2 or 3")
```
