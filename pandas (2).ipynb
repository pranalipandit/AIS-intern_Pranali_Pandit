# Aggregation and Grouping
GroupBy: 
- df.groupby('col'): Groups a DataFrame by one or more columns.

# Aggregation Functions

```python
import pandas as pd

# Create a DataFrame
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Values': [10, 20, 30, 40, 50, 60],
    'Quantity': [1, 2, 1, 2, 1, 2]
}
df = pd.DataFrame(data)
print(df)
```

```python
grouped = df.groupby('Category')
print(grouped.groups)  # Shows the groups and their indices
```

# Sum:- Sum of values in each group.

```python
sum_df = grouped.sum()
print(sum_df)
```

# mean(): Mean of values in each group.

```python
mean_df = grouped.mean()
print(mean_df)
```

# count(): Count of values in each group.

```python
count_df = grouped.count()
print(count_df)
```

```python
# Create a DataFrame
data = {'Category': ['A', 'B', 'A', 'B'], 'Values': [10, 20, 30, 40]}
df = pd.DataFrame(data)

# Group by 'Category' and calculate mean
grouped = df.groupby('Category').mean()
print(grouped)
```

# Merging and Joining:
pd.merge(), df.conacat()

# pd.merge()
- The pd.merge() function is used to combine two DataFrames based on a common column or index.
- Syntax:-pd.merge(left, right, on='key', how='inner')

```python
df1 = pd.DataFrame({'id': [1, 2, 3], 'name': ['A', 'B', 'C']})
df2 = pd.DataFrame({'id': [1, 2, 4], 'age': [25, 31, 42]})
merged_df = pd.merge(df1, df2, on='id')
print(merged_df)
```

# df.conacat()
- The function is used to concatenate DataFrames along a particular axis (rows or columns).
- #syntax:pd.concat([df1, df2], axis=0)

```python
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df_concat = pd.concat([df1, df2])
print(df_concat)
```

# Time Series

- pd.to_datetime():
The pd.to_datetime() function is used to convert a column or series of strings into datetime objects.

- Syntax:pd.to_datetime(df['column'])

```python
# Create a DataFrame with a date column
data = {'Date': ['2025-01-01', '2025-01-02', '2025-01-03'], 'Value': [10, 20, 30]}
df = pd.DataFrame(data)
# Convert 'Date' column to datetime
df['Date'] = pd.to_datetime(df['Date'])
print(df)
```

# resample()
The .resample() method is used to resample time series data (e.g., daily to monthly).

- Syntax: df.resample('rule').aggregation_function()

```python
# Set 'Date' as the index
df.set_index('Date', inplace=True)

# Resample by month and calculate the mean
resampled = df.resample('M').mean()
print(resampled)
```

# Condition Selection
- AND (&) and OR (|) Conditions
- Syntax:
#AND: df[(condition1) & (condition2)]

#OR: df[(condition1) | (condition2)]

# AND :
The & operator is used to combine multiple conditions where all conditions must be true.

```python
# Create a DataFrame
data = {'A': [1, 2, 3, 4], 'B': [10, 20, 30, 40]}
df = pd.DataFrame(data)

# AND condition
filtered_and = df[(df['A'] > 1) & (df['B'] < 40)]
print("AND Condition:")
print(filtered_and)
```

# Or Operator
The | operator is used to combine multiple conditions where at least one condition must be true.

```python
# OR condition
filtered_or = df[(df['A'] > 1) | (df['B'] < 40)]
print("\nOR Condition:")
print(filtered_or)
```
