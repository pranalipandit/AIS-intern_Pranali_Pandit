#### 2. np.zeros()

#### 3.np.ones()
    numpy.ones() is a function in NumPy that creates a new array filled with ones. It allows you to specify the dimensions (shape) of the array by providing the desired shape as an argument or a tuple of dimensions .


```python
arr = np.ones([4,4])
print(arr)
```

```python
arr = np.ones((3, 4))  # 3x4 array of ones
print(arr)
```

#### 4.np.arange()
Generates an array with a sequence of numbers, similar to Python's range(), but returns an array.


```python
arr = np.arange(2, 12, 3)  
print(arr)
```

```python
arr = np.arange(5,9,1)
arr
```

#### 5.np.linspace()

```python
a = np.linspace(2,4)
a
```

```python
a = np.linspace(23,51, retstep=True)                     # retstep : step size bet elements.
a
```

# Array Indexing:

```python
    - Indexing in Numpy refers to accessing and retrieving specific elements or groups of elements from an array. Numpy arrays can be multidimensional,
so indexing allows you to access elements along with different dimensions of the array.
```

```python
arr = np.array ([ 3,6,2,7,2])
print(arr[3])

print(arr[1:5])
```

```python
arr = np.array ([[2,4,1], [5,7,3], [6,4,2]])
print(arr[2,2])

print(arr[0,-2])

print(arr[-2,2])

print(arr[2,0])
```

# slicing 

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[1:4])  # Output: [20, 30, 40] (from index 1 to 3)
print(arr[:3])   # Output: [10, 20, 30] (first 3 elements)
```

```python
arr = np.array([2,6,23,14,5])
print(arr[1:3])

print(arr[-2:4])
```

```python
arr = np.array ([[2,4,1], [5,7,3], [6,4,2]])
print(arr[1:3])

print(arr[-1:-3])               # excluding -3
```
